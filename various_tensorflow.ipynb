# Generadores

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras

# ============================================================
# 1. GENERADOR DE DATOS
# ============================================================
def generate_arrays_from_file(path, batch_size):
  while True:
    inputs = []
    targets = []

    with open(path, "r") as f:
      next(f)  # Saltar cabecera: x,y

      for line in f:
        x, y = line.strip().split(",")
        inputs.append(float(x))
        targets.append(float(y))

        if len(inputs) == batch_size:
          X = np.array(inputs, dtype="float32").reshape(-1, 1)
          y = np.array(targets, dtype="float32")

          yield X, y

          inputs = []
          targets = []

# ============================================================
# 2. CREAR E INSPECCIONAR EL GENERADOR
# ============================================================
batch_size = 5

generator = generate_arrays_from_file("datos.csv", batch_size)

X_batch, y_batch = next(generator)

print("Primer batch X:")
print(X_batch)

print("Primer batch y:")
print(y_batch)

print("Shape X:", X_batch.shape)
print("Shape y:", y_batch.shape)

# ============================================================
# 3. CREAR EL MODELO
# ============================================================
# Crear un modelo secuencial simple con una capa densa de salida única
model = keras.Sequential([
    keras.layers.Input(shape=(1,)),
    keras.layers.Dense(1)
])

# Compilar el modelo con el optimizador Adam y la función de pérdida MSE (error cuadrático medio)
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.01),
    loss="mse"
)

model.summary()

# ============================================================
# 4. INSPECCIONAR PESOS ANTES DE ENTRENAR
# ============================================================
weight_before, bias_before = model.layers[0].get_weights()

print("Weight inicial:", weight_before[0][0])
print("Bias inicial:", bias_before[0])

# ============================================================
# 5. ENTRENAR
# ============================================================
train_generator = generate_arrays_from_file(
    "datos.csv",
    batch_size
)

steps_per_epoch = 30 // batch_size # Número de pasos por época basado en el tamaño del batch y el total de datos (30)

history = model.fit(
    train_generator,
    steps_per_epoch=steps_per_epoch,
    epochs=500,
    shuffle=False,
    verbose=0
)

print("Loss inicial:", history.history["loss"][0])
print("Loss final:", history.history["loss"][-1])

# ============================================================
# 6. INSPECCIONAR LO APRENDIDO
# ============================================================
weight_after, bias_after = model.layers[0].get_weights()

w = weight_after[0][0]
b = bias_after[0]

print("Weight aprendido:", w)
print("Bias aprendido:", b)

print(f"Ecuación aprendida: y ≈ {w:.4f}x + {b:.4f}")
print("Ecuación esperada:  y = 3x + 5")

# ============================================================
# 7. PROBE INDIVIDUAL
# ============================================================
x_probe = np.array([[12.5]], dtype="float32")

prediction = model.predict(x_probe, verbose=0)[0][0]
expected = 3 * 12.5 + 5

print("Probe x=12.5")
print("Predicción:", prediction)
print("Esperado:", expected)

# ============================================================
# 8. TEST CON VARIOS VALORES
# ============================================================
X_test = np.array([
    [2.5],
    [7.5],
    [12.5],
    [18.5],
    [25.5],
    [35.0]
], dtype="float32")

y_test = 3 * X_test.flatten() + 5

predictions = model.predict(
    X_test,
    verbose=0
).flatten()

for x, pred, real in zip(
    X_test.flatten(),
    predictions,
    y_test
):
    error = pred - real

    print(
        f"x={x:5.1f} | "
        f"pred={pred:8.2f} | "
        f"real={real:8.2f} | "
        f"error={error:7.2f}"
    )

# ============================================================
# 9. EVALUAR
# ============================================================
test_loss = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("Test MSE:", test_loss)

# ============================================================
# 10. ERRORES
# ============================================================
errors = predictions - y_test
absolute_errors = np.abs(errors)

print("Errores absolutos:", absolute_errors)
print("Error absoluto medio:", absolute_errors.mean())

# ============================================================
# 11. GRÁFICO DEL LOSS
# ============================================================
plt.figure(figsize=(8, 4))
plt.plot(history.history["loss"])
plt.title("Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.show()

# ============================================================
# 12. RECTA REAL VS MODELO
# ============================================================
x_plot = np.linspace(
    0,
    40,
    100
).reshape(-1, 1)

y_real = 3 * x_plot + 5
y_pred = model.predict(x_plot, verbose=0)

plt.figure(figsize=(8, 5))
plt.plot(
    x_plot,
    y_real,
    label="Real: y = 3x + 5"
)
plt.plot(
    x_plot,
    y_pred,
    linestyle="--",
    label="Modelo"
)
plt.scatter(
    X_test,
    y_test,
    label="Test data"
)

plt.title("Real vs Predicted")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

## Generadores de datos de imágenes con Keras

```python
# ImageDataGenerator - Con Augmentation Layers
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# ============================================================
# 1. CARGAR IMÁGENES DESDE CARPETAS
# ============================================================
train_ds = tf.keras.utils.image_dataset_from_directory(
    "resources",
    image_size=(224, 224),
    batch_size=16,
    label_mode="categorical",
    shuffle=True
)

print(train_ds.class_names)

# ============================================================
# 2. DEFINIR AUGMENTATION
# ============================================================
data_augmentation = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.08),
    layers.RandomTranslation(
        height_factor=0.2,
        width_factor=0.2
    ),
    layers.RandomZoom(0.2),
    layers.RandomContrast(0.2)
])

# ============================================================
# 3. CREAR MODELO
# ============================================================
model = keras.Sequential([
    layers.Input(shape=(224, 224, 3)),

    # Normalización: 0-255 -> 0-1
    layers.Rescaling(1./255),

    # Augmentation
    data_augmentation,

    # Aquí vendrían las capas de la red
    layers.Conv2D(32, 3, activation="relu"),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(2, activation="softmax")
])

# ============================================================
# 4. ENTRENAR
# ============================================================
model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history = model.fit(train_ds, epochs=10)
```

```python
# Continuación del ejemplo con ImageDataGenerator y Augmentation Layers
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

# ============================================================
# 1. VER IMÁGENES ORIGINALES
# ============================================================
class_names = train_ds.class_names

images, labels = next(iter(train_ds))

plt.figure(figsize=(8, 6))

for i in range(min(6, len(images))):
    plt.subplot(2, 3, i + 1)

    plt.imshow(images[i].numpy().astype("uint8"))

    label_index = np.argmax(labels[i])
    plt.title(class_names[label_index])

    plt.axis("off")

plt.suptitle("Original images")
plt.show()


# ============================================================
# 2. VER EL EFECTO DEL AUGMENTATION
# ============================================================
augmented_images = data_augmentation(
    images,
    training=True
)

plt.figure(figsize=(8, 6))

for i in range(min(6, len(augmented_images))):
    plt.subplot(2, 3, i + 1)

    plt.imshow(
        tf.cast(
            tf.clip_by_value(
                augmented_images[i],
                0,
                255
            ),
            tf.uint8
        )
    )

    plt.axis("off")

plt.suptitle("Augmented images")
plt.show()


# ============================================================
# 3. VER CÓMO EVOLUCIONÓ EL ENTRENAMIENTO
# ============================================================
plt.figure(figsize=(8, 4))

plt.plot(
    history.history["loss"],
    label="Loss"
)

plt.title("Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.show()


plt.figure(figsize=(8, 4))

plt.plot(
    history.history["accuracy"],
    label="Accuracy"
)

plt.title("Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.show()


# ============================================================
# 4. CREAR DATASET SIN SHUFFLE PARA PROBAR EL MODELO
# ============================================================
test_ds = tf.keras.utils.image_dataset_from_directory(
    "resources",
    image_size=(224, 224),
    batch_size=12,
    label_mode="categorical",
    shuffle=False
)


# ============================================================
# 5. HACER PREDICCIONES
# ============================================================
test_images, test_labels = next(iter(test_ds))

predictions = model.predict(
    test_images,
    verbose=0
)


# ============================================================
# 6. MOSTRAR PREDICCIONES
# ============================================================
plt.figure(figsize=(10, 8))

for i in range(len(test_images)):
    plt.subplot(3, 4, i + 1)

    plt.imshow(
        test_images[i].numpy().astype("uint8")
    )

    real_index = np.argmax(test_labels[i])
    predicted_index = np.argmax(predictions[i])

    real_class = class_names[real_index]
    predicted_class = class_names[predicted_index]

    confidence = predictions[i][predicted_index] * 100

    plt.title(
        f"Real: {real_class}\n"
        f"Pred: {predicted_class}\n"
        f"{confidence:.1f}%"
    )

    plt.axis("off")

plt.suptitle("Model predictions")
plt.tight_layout()
plt.show()
```

## Generadores de datos custom con Keras

```python
from tensorflow import keras
from pathlib import Path

# ============================================================
# 1. CONFIGURACIÓN
# ============================================================
keras.utils.set_random_seed(42)

IMAGE_SIZE = (128, 128)
BATCH_SIZE = 32
EPOCHS = 20
# Directorio donde se almacenarán los datasets de TensorFlow
DATA_DIR = r"X:\tensorflow_datasets"
```

```python
%pip install -U importlib-resources
import tensorflow_datasets as tfds

# ============================================================
# 2. CARGAR DATASET OFICIAL DE TENSORFLOW
# ============================================================
(train_raw, val_raw, test_raw), info = tfds.load(
    "cats_vs_dogs",
    split=["train[:70%]", "train[70%:85%]", "train[85%:]"],
    as_supervised=True,
    with_info=True,
    data_dir=DATA_DIR
)

class_names = info.features["label"].names

print("Clases:", class_names)
print("Total original:", info.splits["train"].num_examples)
print("Ubicación del dataset:", DATA_DIR)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

# ============================================================
# 3. PREPROCESAR LAS IMÁGENES
# ============================================================
def preprocess(image, label):
    image = tf.image.resize(image, IMAGE_SIZE)
    image = tf.cast(image, tf.float32) / 255.0
    return image, label

# ============================================================
# 4. CREAR PIPELINES POR BATCHES
# ============================================================
AUTOTUNE = tf.data.AUTOTUNE

train_ds = (
    train_raw
    .map(preprocess, num_parallel_calls=AUTOTUNE)
    .shuffle(1000, seed=42)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

val_ds = (
    val_raw
    .map(preprocess, num_parallel_calls=AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

test_ds = (
    test_raw
    .map(preprocess, num_parallel_calls=AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

# ============================================================
# 5. INSPECCIONAR UN BATCH
# ============================================================
X_batch, y_batch = next(iter(train_ds))

print("X shape:", X_batch.shape)
print("y shape:", y_batch.shape)
print("Primeras etiquetas:", y_batch[:8].numpy())

plt.figure(figsize=(10, 5))

for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(X_batch[i].numpy())
    plt.title(class_names[int(y_batch[i])])
    plt.axis("off")

plt.suptitle("Training batch")
plt.tight_layout()
plt.show()

# ============================================================
# 6. DEFINIR DATA AUGMENTATION
# ============================================================
data_augmentation = keras.Sequential([
    keras.layers.RandomFlip("horizontal"),
    keras.layers.RandomRotation(0.10),
    keras.layers.RandomZoom(0.15),
    keras.layers.RandomTranslation(0.10, 0.10),
    keras.layers.RandomContrast(0.15, value_range=(0, 1))
])

# ============================================================
# 7. VISUALIZAR AUGMENTATION
# ============================================================
augmented = data_augmentation(
    X_batch[:4],
    training=True
)

plt.figure(figsize=(10, 5))

for i in range(4):
    plt.subplot(2, 4, i + 1)
    plt.imshow(X_batch[i].numpy())
    plt.title("Original")
    plt.axis("off")

    plt.subplot(2, 4, i + 5)
    plt.imshow(np.clip(augmented[i].numpy(), 0, 1))
    plt.title("Augmented")
    plt.axis("off")

plt.tight_layout()
plt.show()

# ============================================================
# 8. CREAR MODELO CNN
# ============================================================
model = keras.Sequential([
    keras.layers.Input(shape=(128, 128, 3)),
    data_augmentation,

    keras.layers.Conv2D(16, 3, activation="relu"),
    keras.layers.MaxPooling2D(),

    keras.layers.Conv2D(32, 3, activation="relu"),
    keras.layers.MaxPooling2D(),

    keras.layers.Conv2D(64, 3, activation="relu"),
    keras.layers.MaxPooling2D(),

    keras.layers.GlobalAveragePooling2D(),
    keras.layers.Dense(32, activation="relu"),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(2, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# ============================================================
# 9. ENTRENAR
# ============================================================
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=[
        keras.callbacks.EarlyStopping(
            monitor="val_loss",
            patience=3,
            restore_best_weights=True
        )
    ]
)

# ============================================================
# 10. VISUALIZAR EVOLUCIÓN DEL LOSS
# ============================================================
plt.figure(figsize=(8, 4))
plt.plot(history.history["loss"], label="Training")
plt.plot(history.history["val_loss"], label="Validation")
plt.title("Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.show()

# ============================================================
# 11. VISUALIZAR EVOLUCIÓN DEL ACCURACY
# ============================================================
plt.figure(figsize=(8, 4))
plt.plot(history.history["accuracy"], label="Training")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.title("Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.show()

# ============================================================
# 12. EVALUAR CON TEST
# ============================================================
test_loss, test_accuracy = model.evaluate(
    test_ds,
    verbose=1
)

print(f"Test loss: {test_loss:.4f}")
print(f"Test accuracy: {test_accuracy:.2%}")

# ============================================================
# 13. PREDECIR IMÁGENES DEL TEST
# ============================================================
X_test, y_test = next(iter(test_ds))

predictions = model.predict(
    X_test[:12],
    verbose=0
)

plt.figure(figsize=(12, 9))

for i in range(12):
    real = int(y_test[i])
    predicted = int(np.argmax(predictions[i]))
    confidence = float(predictions[i][predicted]) * 100

    plt.subplot(3, 4, i + 1)
    plt.imshow(X_test[i].numpy())
    plt.title(
        f"Real: {class_names[real]}\n"
        f"Pred: {class_names[predicted]}\n"
        f"{confidence:.1f}%"
    )
    plt.axis("off")

plt.suptitle("Test predictions")
plt.tight_layout()
plt.show()

# ============================================================
# 14. PROBAR UNA IMAGEN EXTERNA
# ============================================================
def predict_image(model, path):
    image = keras.utils.load_img(
        path,
        target_size=IMAGE_SIZE
    )
    image_array = keras.utils.img_to_array(image) / 255.0
    input_data = np.expand_dims(image_array, axis=0)

    prediction = model.predict(input_data, verbose=0)[0]
    predicted_index = int(np.argmax(prediction))
    confidence = float(prediction[predicted_index]) * 100

    print("Predicción:", class_names[predicted_index])
    print(f"Confianza: {confidence:.2f}%")

    plt.imshow(image)
    plt.title(f"{class_names[predicted_index]} ({confidence:.1f}%)")
    plt.axis("off")
    plt.show()
```
