```python
x=np.linspace(0,2,100)

fig, ax =plt.subplots(figsize=(10,6), layout='constrained')
ax.plot(x,x, label='linear')
ax.plot(x,x ** 2, label='quadratic')
ax.plot(x,x ** 3, label='cubic', marker="o", markevery=20)

# Punto donde coinciden las tres funciones
x_point = 1
y_point = 1

# Marcar el punto
ax.scatter(x_point, y_point)

# Líneas de proyección hacia los ejes
ax.vlines(
    x=x_point,
    ymin=0,
    ymax=y_point,
    linestyles="dashed"
)

ax.hlines(
    y=y_point,
    xmin=0,
    xmax=x_point,
    linestyles="dashed"
)

ax.set_xlabel("x label")
ax.set_ylabel("y label")
ax.set_title("Simple Plot")
ax.legend()

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Datos para gráficos categóricos
ventas = pd.Series(
    [120, 200, 150, 250],
    index=["Enero", "Febrero", "Marzo", "Abril"]
)

# Datos para estudiar una distribución
np.random.seed(42)
datos = pd.Series(np.random.normal(100, 15, 500))

# ============================================================
# LINE
# Evolución o tendencia
# ============================================================
ventas.plot(kind="line", marker="o", title="Ventas por mes")
plt.show()

# ============================================================
# BAR
# Comparación entre categorías
# ============================================================
ventas.plot(kind="bar", title="Ventas por mes")
plt.show()

# ============================================================
# BARH
# Comparación horizontal entre categorías
# ============================================================
ventas.plot(kind="barh", title="Ventas por mes")
plt.show()

# ============================================================
# PIE
# Participación de cada elemento sobre el total
# ============================================================
ventas.plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Distribución de ventas"
)

plt.ylabel("")
plt.show()

# ============================================================
# HIST
# Frecuencia de valores numéricos por intervalos
# ============================================================
datos.plot(
    kind="hist",
    bins=20,
    title="Distribución de valores"
)
plt.show()

# ============================================================
# AREA
# Evolución destacando el área bajo la curva
# ============================================================
ventas.plot(
    kind="area",
    alpha=0.5,
    title="Evolución de ventas"
)
plt.show()

# ============================================================
# KDE / DENSITY
# Estimación suavizada de una distribución
# ============================================================
datos.plot(
    kind="kde",
    title="Densidad de los valores"
)
plt.show()
```

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([10, 20, 30])

# ============================================================
# GUARDAR UN ARRAY
# ============================================================
np.save("array.npy", arr1)


# ============================================================
# CARGAR UN ARRAY
# ============================================================
loaded = np.load("array.npy")
print(loaded)
# [1 2 3]

# ============================================================
# GUARDAR VARIOS ARRAYS
# ============================================================
np.savez(
    "arrays.npz",
    primero=arr1,
    segundo=arr2
)

# ============================================================
# CARGAR VARIOS ARRAYS
# ============================================================
loaded = np.load("arrays.npz")

print(loaded["primero"])
# [1 2 3]

print(loaded["segundo"])
# [10 20 30]

# ============================================================
# GUARDAR VARIOS ARRAYS COMPRIMIDOS
# ============================================================
np.savez_compressed(
    "arrays_comprimidos.npz",
    primero=arr1,
    segundo=arr2
)
```

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Luis", "Marta", "Pedro"],
    "edad": [25, np.nan, np.nan, 42, 71],
    "ciudad": ["Madrid", "Bogota", "Bogota", "MADRID", "Roma"],
    "salario": [30000, 40000, np.nan, 50000, None]
})
display(df)

# drop na
nonulldf = df.dropna()
display(nonulldf)
# Output
#    nombre  edad  ciudad  salario
# 0    Ana  25.0  Madrid  30000.0
# 1    Marta  42.0  MADRID  50000.0


# fill na
filleddf = df.fillna({
  "edad": df["edad"].mean(),
  "salario": 0.
  })
display(filleddf)
# Output
#    nombre edad  ciudad  salario
# 0    Ana  25  Madrid  30000.0
# 1    Luis  46  Bogota  40000.0
# 2    Luis  46  Bogota      0.0
# 3   Marta  42  MADRID  50000.0
# 4   Pedro  71    Roma      0.0

# forward fill and backward fill demonstration
df = pd.DataFrame(
  [
    [np.nan, 2, np.nan, 0, -1],
    [3, 4, np.nan, 1, 10],
    [np.nan, np.nan, np.nan, np.nan, np.pi],
    [np.nan, 3, np.nan, 4, np.nan],
  ],
  columns=list("ABCDE"),
)
display(df)
# Output
#    A    B    C  D
# 0  NaN  2.0  NaN  0
# 1  3.0  4.0  NaN  1
# 2  NaN  NaN  NaN  NaN
# 3  NaN  3.0  NaN  4

forwardfill = df.ffill()
display(forwardfill)

# Output
#    A    B    C  D
# 0  NaN  2.0  NaN  0
# 1  3.0  4.0  NaN  1
# 2  3.0  4.0  NaN  1
# 3  3.0  3.0  NaN  4

backwardfill = df.bfill()
display(backwardfill)

# Output
#    A    B    C  D
# 0  3.0  2.0  NaN  0
# 1  3.0  4.0  NaN  1
# 2  NaN  3.0  NaN  4
# 3  NaN  3.0  NaN  4


# Interpolate dataframe
interpolated = df.interpolate()
display(interpolated)

# Output
#    A    B    C  D    E
# 0  NaN  2.0  NaN  0  -1.0
# 1  3.0  4.0  NaN  1  10.0
# 2  NaN  NaN  NaN  NaN  3.141593
# 3  NaN  3.0  NaN  4  NaN

```

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Luis", "Marta", "Pedro"],
    "edad": [25, np.nan, np.nan, 42, 71],
    "ciudad": ["Madrid", "Bogota", "Bogota", "MADRID", "Roma"]
})


# ============================================================
# 1. DETECTAR VALORES AUSENTES
# ============================================================
print(df.isna())
print(df.isna().sum())


# ============================================================
# 2. RELLENAR NaN
# ============================================================
df["edad"] = df["edad"].fillna(
    df["edad"].median()
)
print(df)


# ============================================================
# 3. DETECTAR DUPLICADOS
# ============================================================
print(df.duplicated(
  subset=["edad"]
))


# ============================================================
# 4. ELIMINAR DUPLICADOS
# ============================================================
df = df.drop_duplicates()
print(df)


# ============================================================
# 5. NORMALIZAR VALORES
# ============================================================
df["ciudad"] = df["ciudad"].replace({
    "MADRID": "Madrid",
    "Bogota": "Bogotá"
})


# ============================================================
# 6. CONVERTIR EDAD EN CATEGORÍA
# ============================================================
df["grupo_edad"] = pd.cut(
    df["edad"],
    bins=[0, 18, 35, 65, 100],
    labels=["Menor", "Joven", "Adulto", "Mayor"]
)

# ============================================================
# 7. Categorias con qcut
# ============================================================
df["grupo_edad_qcut"] = pd.qcut(
    df["edad"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)
print(df)

# ============================================================
# 8. CONTAR CATEGORÍAS
# ============================================================
print(df["grupo_edad"].value_counts())
```

```python
# ============================================================
# FUNCIÓN GENERADORA
# ============================================================
def cuadrados(n):
    for i in range(n):
        yield i ** 2
g = cuadrados(5)

# ============================================================
# OBTENER VALORES MANUALMENTE
# ============================================================
print(next(g))
# 0
print(next(g))
# 1

# ============================================================
# CONTINUAR ITERANDO
# ============================================================
for valor in g:
    print(valor)
# 4
# 9
# 16

# ============================================================
# LIST COMPREHENSION
# Crea todos los valores inmediatamente
# ============================================================
lista = [x ** 2 for x in range(5)]
print(lista)
# [0, 1, 4, 9, 16]

# ============================================================
# GENERATOR EXPRESSION
# Produce los valores bajo demanda
# ============================================================
generador = (x ** 2 for x in range(5))

print(next(generador))
# 0
print(next(generador))
# 1
```
