```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Datos para gráficos categóricos
ventas = pd.Series(
    [120, 200, 150, 250],
    index=["Enero", "Febrero", "Marzo", "Abril"]
)

# Datos para estudiar una distribución
np.random.seed(42)
datos = pd.Series(np.random.normal(100, 15, 500))


# ============================================================
# LINE
# Evolución o tendencia
# ============================================================

ventas.plot(kind="line", marker="o", title="Ventas por mes")
plt.show()


# ============================================================
# BAR
# Comparación entre categorías
# ============================================================

ventas.plot(kind="bar", title="Ventas por mes")
plt.show()


# ============================================================
# BARH
# Comparación horizontal entre categorías
# ============================================================

ventas.plot(kind="barh", title="Ventas por mes")
plt.show()


# ============================================================
# PIE
# Participación de cada elemento sobre el total
# ============================================================

ventas.plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Distribución de ventas"
)

plt.ylabel("")
plt.show()


# ============================================================
# HIST
# Frecuencia de valores numéricos por intervalos
# ============================================================

datos.plot(
    kind="hist",
    bins=20,
    title="Distribución de valores"
)

plt.show()


# ============================================================
# AREA
# Evolución destacando el área bajo la curva
# ============================================================

ventas.plot(
    kind="area",
    alpha=0.5,
    title="Evolución de ventas"
)

plt.show()


# ============================================================
# KDE / DENSITY
# Estimación suavizada de una distribución
# ============================================================

datos.plot(
    kind="kde",
    title="Densidad de los valores"
)

plt.show()
```

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Luis", "Marta", "Pedro"],
    "edad": [25, np.nan, np.nan, 42, 71],
    "ciudad": ["Madrid", "Bogota", "Bogota", "MADRID", "Roma"]
})


# ============================================================
# 1. DETECTAR VALORES AUSENTES
# ============================================================

print(df.isna())
print(df.isna().sum())


# ============================================================
# 2. RELLENAR NaN
# ============================================================

df["edad"] = df["edad"].fillna(
    df["edad"].median()
)
print(df)


# ============================================================
# 3. DETECTAR DUPLICADOS
# ============================================================

print(df.duplicated(
  subset=["edad"]
))


# ============================================================
# 4. ELIMINAR DUPLICADOS
# ============================================================

df = df.drop_duplicates()
print(df)


# ============================================================
# 5. NORMALIZAR VALORES
# ============================================================

df["ciudad"] = df["ciudad"].replace({
    "MADRID": "Madrid",
    "Bogota": "Bogotá"
})


# ============================================================
# 6. CONVERTIR EDAD EN CATEGORÍA
# ============================================================

df["grupo_edad"] = pd.cut(
    df["edad"],
    bins=[0, 18, 35, 65, 100],
    labels=["Menor", "Joven", "Adulto", "Mayor"]
)

# ============================================================
# 7. Categorias con qcut
# ============================================================
df["grupo_edad_qcut"] = pd.qcut(
    df["edad"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)
print(df)

# ============================================================
# 8. CONTAR CATEGORÍAS
# ============================================================

print(df["grupo_edad"].value_counts())
```

```python
# ============================================================
# FUNCIÓN GENERADORA
# ============================================================
def cuadrados(n):
    for i in range(n):
        yield i ** 2

g = cuadrados(5)

# ============================================================
# OBTENER VALORES MANUALMENTE
# ============================================================
print(next(g))
# 0

print(next(g))
# 1

# ============================================================
# CONTINUAR ITERANDO
# ============================================================
for valor in g:
    print(valor)

# 4
# 9
# 16

# ============================================================
# LIST COMPREHENSION
# Crea todos los valores inmediatamente
# ============================================================
lista = [x ** 2 for x in range(5)]

print(lista)
# [0, 1, 4, 9, 16]

# ============================================================
# GENERATOR EXPRESSION
# Produce los valores bajo demanda
# ============================================================
generador = (x ** 2 for x in range(5))

print(next(generador))
# 0

print(next(generador))
# 1
```

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras

# ============================================================
# 1. GENERADOR DE DATOS
# ============================================================
def generate_arrays_from_file(path, batch_size):
  while True:
    inputs = []
    targets = []

    with open(path, "r") as f:
      next(f)  # Saltar cabecera: x,y

      for line in f:
        x, y = line.strip().split(",")
        inputs.append(float(x))
        targets.append(float(y))

        if len(inputs) == batch_size:
          X = np.array(inputs, dtype="float32").reshape(-1, 1)
          y = np.array(targets, dtype="float32")

          yield X, y

          inputs = []
          targets = []

# ============================================================
# 2. CREAR E INSPECCIONAR EL GENERADOR
# ============================================================
batch_size = 5

generator = generate_arrays_from_file("datos.csv", batch_size)

X_batch, y_batch = next(generator)

print("Primer batch X:")
print(X_batch)

print("Primer batch y:")
print(y_batch)

print("Shape X:", X_batch.shape)
print("Shape y:", y_batch.shape)

# ============================================================
# 3. CREAR EL MODELO
# ============================================================
# Crear un modelo secuencial simple con una capa densa de salida única
model = keras.Sequential([
    keras.layers.Input(shape=(1,)),
    keras.layers.Dense(1)
])

# Compilar el modelo con el optimizador Adam y la función de pérdida MSE (error cuadrático medio)
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.01),
    loss="mse"
)

model.summary()

# ============================================================
# 4. INSPECCIONAR PESOS ANTES DE ENTRENAR
# ============================================================
weight_before, bias_before = model.layers[0].get_weights()

print("Weight inicial:", weight_before[0][0])
print("Bias inicial:", bias_before[0])

# ============================================================
# 5. ENTRENAR
# ============================================================
train_generator = generate_arrays_from_file(
    "datos.csv",
    batch_size
)

steps_per_epoch = 30 // batch_size # Número de pasos por época basado en el tamaño del batch y el total de datos (30)

history = model.fit(
    train_generator,
    steps_per_epoch=steps_per_epoch,
    epochs=500,
    shuffle=False,
    verbose=0
)

print("Loss inicial:", history.history["loss"][0])
print("Loss final:", history.history["loss"][-1])

# ============================================================
# 6. INSPECCIONAR LO APRENDIDO
# ============================================================
weight_after, bias_after = model.layers[0].get_weights()

w = weight_after[0][0]
b = bias_after[0]

print("Weight aprendido:", w)
print("Bias aprendido:", b)

print(f"Ecuación aprendida: y ≈ {w:.4f}x + {b:.4f}")
print("Ecuación esperada:  y = 3x + 5")

# ============================================================
# 7. PROBE INDIVIDUAL
# ============================================================
x_probe = np.array([[12.5]], dtype="float32")

prediction = model.predict(x_probe, verbose=0)[0][0]
expected = 3 * 12.5 + 5

print("Probe x=12.5")
print("Predicción:", prediction)
print("Esperado:", expected)

# ============================================================
# 8. TEST CON VARIOS VALORES
# ============================================================
X_test = np.array([
    [2.5],
    [7.5],
    [12.5],
    [18.5],
    [25.5],
    [35.0]
], dtype="float32")

y_test = 3 * X_test.flatten() + 5

predictions = model.predict(
    X_test,
    verbose=0
).flatten()

for x, pred, real in zip(
    X_test.flatten(),
    predictions,
    y_test
):
    error = pred - real

    print(
        f"x={x:5.1f} | "
        f"pred={pred:8.2f} | "
        f"real={real:8.2f} | "
        f"error={error:7.2f}"
    )

# ============================================================
# 9. EVALUAR
# ============================================================
test_loss = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("Test MSE:", test_loss)

# ============================================================
# 10. ERRORES
# ============================================================
errors = predictions - y_test
absolute_errors = np.abs(errors)

print("Errores absolutos:", absolute_errors)
print("Error absoluto medio:", absolute_errors.mean())

# ============================================================
# 11. GRÁFICO DEL LOSS
# ============================================================
plt.figure(figsize=(8, 4))
plt.plot(history.history["loss"])
plt.title("Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.show()

# ============================================================
# 12. RECTA REAL VS MODELO
# ============================================================
x_plot = np.linspace(
    0,
    40,
    100
).reshape(-1, 1)

y_real = 3 * x_plot + 5
y_pred = model.predict(x_plot, verbose=0)

plt.figure(figsize=(8, 5))
plt.plot(
    x_plot,
    y_real,
    label="Real: y = 3x + 5"
)
plt.plot(
    x_plot,
    y_pred,
    linestyle="--",
    label="Modelo"
)
plt.scatter(
    X_test,
    y_test,
    label="Test data"
)

plt.title("Real vs Predicted")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
